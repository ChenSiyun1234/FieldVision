```python
CHECK_SCRIPT = '\n"""Same-result check: re-extract finished laptop blocks and compare exactly."""\nimport json, shutil, sys, time, zipfile\nfrom argparse import Namespace\nfrom concurrent.futures import ThreadPoolExecutor\nfrom pathlib import Path\nimport numpy as np\nimport pandas as pd\nfrom fieldvision.s2.core import BANDS\nfrom fieldvision.s2.extract import process_block\nfrom fieldvision.s2.geometry import load_fields, make_blocks\n\nwork, check_zip, workers, main_fields = Path(sys.argv[1]), Path(sys.argv[2]), int(sys.argv[3]), Path(sys.argv[4])\nblock_ids = sys.argv[5:]\nroot = work / "check"\nshutil.rmtree(root, ignore_errors=True)\nzipfile.ZipFile(check_zip).extractall(work)\nEXACT = ["scene_id", "collection", "n_pixels", "n_clear", "clear_fraction", *BANDS]\nfailed = False\nwith ThreadPoolExecutor(max_workers=workers) as pool:\n    for block_id in block_ids:\n        d = root / block_id\n        s = json.loads((d / "settings.json").read_text())\n        p = s["parameters"]\n        fields = main_fields if s.get("fields") == "MAIN" else d / "fields.geojson"\n        blocks = {b["id"]: b for b in make_blocks(load_fields(fields, s["year"]), p["block_size"])}\n        block = blocks[block_id]\n        out = d / "out"\n        (out / "blocks").mkdir(parents=True)\n        (out / "catalog").mkdir()\n        shutil.copy(d / "catalog.json", out / "catalog" / f"{block_id}.json")\n        for attempt in range(4):\n            status = process_block(block, Namespace(**p, out=out, scene_batch=6), pool)\n            errors = status.get("scene_errors", {})\n            if status["status"] == "complete" or not all("RasterioIOError" in e or "CPLE_" in e for e in errors.values()):\n                break\n            print(f"{block_id}: {len(errors)} scene reads failed (network), retry {attempt + 1}", flush=True)\n            time.sleep(30 * (attempt + 1))\n        old = pd.read_parquet(d / "reference.parquet").set_index(["field_id", "date"]).sort_index()\n        new = pd.read_parquet(out / "blocks" / f"{block_id}.parquet").set_index(["field_id", "date"]).sort_index()\n        problems = []\n        if status["status"] != "complete":\n            problems.append(f"block status {status[\'status\']}: {status.get(\'scene_errors\')}")\n        if not old.index.equals(new.index):\n            problems.append(f"rows differ: laptop {len(old)}, here {len(new)}, "\n                            f"{len(old.index.symmetric_difference(new.index))} keys not shared")\n            common = old.index.intersection(new.index)\n            old, new = old.loc[common], new.loc[common]\n        for column in EXACT:\n            a, b = old[column].to_numpy(), new[column].to_numpy()\n            same = (a == b) | (pd.isna(a) & pd.isna(b)) if a.dtype.kind == "f" else (a == b)\n            if not np.all(same):\n                first = np.flatnonzero(~same)[0]\n                problems.append(f"{column}: {int((~same).sum())} rows differ, e.g. {old.index[first]}: laptop {a[first]!r}, here {b[first]!r}")\n        print(f"{block_id} ({s[\'source_run\']}): {len(old)} rows, {status[\'scenes_succeeded\']} scenes -> "\n              + ("IDENTICAL" if not problems else "DIFFERENT"))\n        for problem in problems:\n            print("   ", problem)\n        failed |= bool(problems)\nprint("CHECK", "FAILED" if failed else "PASSED")\nsys.exit(1 if failed else 0)\n'
# 1. Settings (only WORKERS may be changed; it does not change results)
import os, sys, json, time, shutil, zipfile, hashlib, subprocess, threading
from pathlib import Path
UPLOAD = Path("/content")
WORK = Path("/content/fv")
bundle = json.loads((UPLOAD / "bundle.json").read_text())
RUN = bundle["run"]
OUT = WORK / RUN
CODE = WORK / "code"
DRIVE_BACKUP = Path("/content/drive/MyDrive/fieldvision_" + RUN)
CPUS = os.cpu_count() or 2
WORKERS = max(16, 8 * CPUS)   # I/O threads for (scene, band) reads
print(f"{RUN}: {CPUS} CPUs, {WORKERS} read threads; Python {sys.version.split()[0]}")
needed = [bundle["code_zip"], "corn_grain_silage_acres_2012_2025.csv", "corn_grain_silage_acres_2021_2023.csv", "check_blocks.zip", *bundle["catalog_parts"], *bundle["block_parts"]]
missing = [n for n in needed if not (UPLOAD / n).exists()]
print("missing uploads:", missing or "none")
```

```python
# 4. Unpack the frozen code, check its fingerprint, then the same-result check
shutil.rmtree(CODE, ignore_errors=True)
zipfile.ZipFile(UPLOAD / bundle["code_zip"]).extractall(CODE)
digest = hashlib.sha256(b"".join(p.read_bytes() for p in sorted((CODE / "fieldvision" / "s2").glob("*.py")))).hexdigest()
print("code fingerprint", digest)
assert digest == bundle["code_sha256"], "code fingerprint differs from the laptop run's code"
WORK.mkdir(parents=True, exist_ok=True)
(WORK / "check.py").write_text(CHECK_SCRIPT)
env = dict(os.environ, PYTHONPATH=str(CODE))
passed_marker = WORK / "check_passed.txt"
stamp = digest + "\n" + versions
if passed_marker.exists() and passed_marker.read_text() == stamp:
    print("same-result check already PASSED in this runtime with this code and these libraries; not repeated")
else:
    result = subprocess.run([sys.executable, "-W", "ignore", str(WORK / "check.py"), str(WORK), str(UPLOAD / "check_blocks.zip"),
                             str(min(WORKERS, 16)), str(UPLOAD / bundle["fields"]), *bundle["check_blocks"]],
                            env=env, capture_output=True, text=True)
    print(result.stdout[-6000:])
    if result.returncode != 0:
        print(result.stderr[-3000:])
        raise SystemExit("Same-result check FAILED: do not use this runtime for the extraction. Tell the coordinator.")
    passed_marker.write_text(stamp)
```

```python
# A. National CSB archive by range requests (a plain GET returns an empty body) and the state subset.
#    Needs geopandas/pyogrio (installed without dependencies, so the pinned versions above stay).
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--only-binary=:all:", "--no-deps",
                "geopandas==1.2.0", "pyogrio==0.13.0", "packaging", "certifi"], check=True)
FIPS, YEAR, UTM, QS = "55101", 2023, "EPSG:32616", UPLOAD / "corn_grain_silage_acres_2012_2025.csv"
FIELDS_OUT = Path("/content/out")          # outside the code tree: from_file refuses folders inside it
FIELDS_OUT.mkdir(parents=True, exist_ok=True)
CSB_ZIP = Path("/content/csb/NationalCSB_2018-2025_rev23.zip")
STATE_PQ = FIELDS_OUT / "csb1825_55.parquet"

def run_py(*args):
    print("$", " ".join(map(str, args)), flush=True)
    r = subprocess.run([sys.executable, "-W", "ignore", *map(str, args)], env=env, capture_output=True, text=True)
    print(r.stdout[-3000:])
    if r.returncode != 0:
        print(r.stderr[-3000:])
        raise SystemExit("failed: " + str(args[0]))

if not STATE_PQ.exists():
    CSB_ZIP.parent.mkdir(parents=True, exist_ok=True)
    if not CSB_ZIP.exists() or CSB_ZIP.stat().st_size != 3758023527 or Path(str(CSB_ZIP) + ".done").exists():
        run_py(CODE / "scripts" / "range_download.py", "https://www.nass.usda.gov/sites/default/files/Research_and_Science/Crop-Sequence-Boundaries/datasets/NationalCSB_2018-2025_rev23.zip", CSB_ZIP, "--workers", "8")
    assert CSB_ZIP.stat().st_size == 3758023527
    run_py(CODE / "scripts" / "csb_subset.py", "--zip", CSB_ZIP, "--state", "55", "--out", STATE_PQ)
    CSB_ZIP.unlink()
print(json.loads(STATE_PQ.with_suffix(".json").read_text())["rows"], "CSB polygons in state 55")
```

```python
# B. The county's corn fields (CSB polygons whose CDL2023 majority class is corn) and the corn-area check
TIGER = Path("/content/tl_2022_us_county.zip")
if not TIGER.exists():
    subprocess.run(["curl", "-sS", "-L", "--retry", "3", "-o", str(TIGER), "https://www2.census.gov/geo/tiger/TIGER2022/COUNTY/tl_2022_us_county.zip"], check=True)
COUNTY_DIR = FIELDS_OUT / RUN
if not (COUNTY_DIR / "fields.geojson").exists():
    run_py("-m", "fieldvision.labels.from_file", STATE_PQ, "--where", "STATEFIPS=55", "--where", "CNTYFIPS=101",
           "--where", "CDL2023=1", "--unlabelled", "--year", "2023", "--id-column", "CSBID", "--no-year-in-id",
           "--prefix", "csb", "--country", "US", "--source", "nass-csb", "--crop-code-column", "CDL2023",
           "--crop-name", "Corn", "--metric-crs", UTM, "--keep", "CSBACRES=acres_reported",
           "--set", "county_fips=55101", "--out", COUNTY_DIR)
run_py(CODE / "scripts" / "county_corn_check.py", "--fields", COUNTY_DIR / "fields.geojson", "--fips", FIPS, "--year", "2023",
       "--census-year", "2022", "--survey-csv", UPLOAD / "corn_grain_silage_acres_2021_2023.csv", "--survey-year", "2023", "--counties", TIGER, "--quickstats", QS, "--csb", STATE_PQ,
       "--utm", UTM, "--out", COUNTY_DIR / "county_check.json")
c = json.loads((COUNTY_DIR / "county_check.json").read_text())
s = c.get("survey") or {}
print(f"{RUN}: {c['fields']} fields, {c['area_acres_total']:,.0f} corn acres in CSB polygons")
print(f"  Census 2022: grain {c['usda']['grain_harvested_acres']}, silage {c['usda']['silage_harvested_acres']} "
      f"-> polygons / (grain+silage) = {c.get('polygon_acres_over_census_harvested')}")
print(f"  Survey 2023: planted {s.get('planted_acres')}, grain {s.get('grain_harvested_acres')}, silage {s.get('silage_harvested_acres')} "
      f"-> polygons / (grain+silage) = {c.get('polygon_acres_over_survey_harvested')}, / planted = {c.get('polygon_acres_over_survey_planted')}")
```

```python
# S. Cap the extraction at about 70 read blocks by drawing whole 10 km blocks (fixed seed); record the draw
import numpy as np
sys.path.insert(0, str(CODE))
from fieldvision.s2.geometry import load_fields, make_blocks
MAX_BLOCKS, SEED, ACRES_PER_HA = 70, 42, 2.471053814671653
full = json.loads((COUNTY_DIR / "fields.geojson").read_text(encoding="utf-8"))
read_blocks = make_blocks(load_fields(COUNTY_DIR / "fields.geojson", YEAR), 6000)
per_label = {}
for blk in read_blocks:
    lb = blk["fields"][0]["properties"]["block_id"]
    per_label[lb] = per_label.get(lb, 0) + 1
labels = sorted(per_label)
order = [labels[i] for i in np.random.default_rng(SEED).permutation(len(labels))]
drawn, used = [], 0
for lb in order:
    if used + per_label[lb] <= MAX_BLOCKS:
        drawn.append(lb)
        used += per_label[lb]
keep = set(drawn)
subset = dict(full, features=[f for f in full["features"] if f["properties"]["block_id"] in keep])
area = lambda feats: sum(f["properties"]["area_ha"] for f in feats) * ACRES_PER_HA
sample = {"run": RUN, "rule": "whole 10 km blocks (block_id) in numpy default_rng(seed).permutation order of the sorted ids; "
                          "a block is kept if its read blocks still fit under max_read_blocks",
          "seed": SEED, "max_read_blocks": MAX_BLOCKS, "is_sample": len(drawn) < len(labels),
          "label_blocks_drawn": len(drawn), "label_blocks_total": len(labels),
          "read_blocks_drawn": used, "read_blocks_total": len(read_blocks),
          "fields_drawn": len(subset["features"]), "fields_total": len(full["features"]),
          "acres_drawn": round(area(subset["features"]), 1), "acres_total": round(area(full["features"]), 1),
          "drawn_label_blocks": sorted(drawn)}
sample["fraction_acres"] = round(sample["acres_drawn"] / sample["acres_total"], 4)
(COUNTY_DIR / "sample.json").write_text(json.dumps(sample, indent=2))
(UPLOAD / bundle["fields"]).write_text(json.dumps(subset, ensure_ascii=False), encoding="utf-8")   # what the extraction reads
bundle["block_count"] = len(make_blocks(load_fields(UPLOAD / bundle["fields"], YEAR), 6000))
assert bundle["block_count"] == used, (bundle["block_count"], used)
(UPLOAD / "bundle.json").write_text(json.dumps(bundle, indent=2))
print({k: v for k, v in sample.items() if k != "drawn_label_blocks"})
```

```python
# 5. Prepare the output folder: restore from Drive, else unpack the laptop checkpoints; add the catalog cache
OUT.mkdir(parents=True, exist_ok=True)
if DRIVE_BACKUP.exists() and not (OUT / "blocks").exists():
    shutil.copytree(DRIVE_BACKUP, OUT, dirs_exist_ok=True)
    print("restored checkpoints from Drive")

def unpack(parts, name):
    if not parts:
        return 0
    (WORK / name).write_bytes(b"".join((UPLOAD / n).read_bytes() for n in parts))
    added = 0
    with zipfile.ZipFile(WORK / name) as archive:
        for member in archive.namelist():
            target = OUT / member
            if not target.exists():          # never overwrite work done here
                target.parent.mkdir(parents=True, exist_ok=True)
                target.write_bytes(archive.read(member))
                added += 1
    return added

print("catalog files added:", unpack(bundle["catalog_parts"], "catalog.zip"))
print("laptop block files added:", unpack(bundle["block_parts"], "blocks.zip"))
start_state = WORK / "start_state.json"
if not start_state.exists():
    state = {p.stem: json.loads(p.read_text()).get("status") for p in (OUT / "blocks").glob("*.json")}
    start_state.write_text(json.dumps(state))
state = json.loads(start_state.read_text())
print("blocks at start:", {s: list(state.values()).count(s) for s in set(state.values())}, "of", bundle["block_count"])
```

```python
# 6. Run the extraction (same arguments as the laptop run). Rerun this cell to resume.
cmd = [sys.executable, "-W", "ignore", "-m", "fieldvision.s2.extract", "--fields", str(UPLOAD / bundle["fields"]),
       *bundle["cli"], "--out", str(OUT), "--workers", str(WORKERS), "--scene-batch", "6"]
print(" ".join(cmd))
stop = threading.Event()

def backup():
    while not stop.wait(300):
        if Path("/content/drive/MyDrive").exists():
            shutil.copytree(OUT / "blocks", DRIVE_BACKUP / "blocks", dirs_exist_ok=True, ignore=shutil.ignore_patterns("*.tmp"))
            if (OUT / "manifest.json").exists():
                shutil.copy2(OUT / "manifest.json", DRIVE_BACKUP / "manifest.json")

def finished():
    return sum(json.loads(p.read_text()).get("status") in ("complete", "partial") for p in (OUT / "blocks").glob("*.json"))

threading.Thread(target=backup, daemon=True).start()
started = time.time()
with open(OUT / "run.log", "a") as log:
    proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in proc.stdout:
        log.write(line)
        if "boto3" in line:
            continue
        if " fields, " in line or " ERROR " in line or "complete:" in line or "partial:" in line:
            print(f"[{time.strftime('%H:%M:%S')}] {finished()}/{bundle['block_count']} blocks done | {line.strip()[24:]}", flush=True)
    code_ = proc.wait()
stop.set()
print("exit code", code_, "(0 complete, 2 partial)", f"{(time.time()-started)/60:.1f} min this session")
failures = {}
for p in (OUT / "blocks").glob("*.json"):
    for e in json.loads(p.read_text()).get("scene_errors", {}).values():
        kind = "network read (rerun this cell)" if ("RasterioIOError" in e or "CPLE_" in e) else e[:60]
        failures[kind] = failures.get(kind, 0) + 1
print("scene errors by kind:", failures or "none")
```

```python
# 7. Results: only blocks worked on here (parquet + checkpoint json) and the manifest, zipped,
#    split into parts of at most 50 MB with SHA256SUMS.txt. Download every part and the checksum file.
manifest = json.loads((OUT / "manifest.json").read_text())
print("status", manifest["status"], "| blocks", {b["status"] for b in manifest["blocks"].values()},
      "| scene failures", manifest["scene_failures"], "| fields with usable dates", manifest.get("fields_with_usable_dates"))
changed = []
for p in sorted((OUT / "blocks").glob("*.json")):
    now = json.loads(p.read_text()).get("status")
    if state.get(p.stem) != "complete" and now in ("complete", "partial"):
        changed.append(p.stem)
print(len(changed), "blocks to send back; still unfinished:",
      [p.stem for p in (OUT / "blocks").glob("*.json") if json.loads(p.read_text()).get("status") == "running"])
result = Path(f"/content/{RUN}_colab_blocks.zip")
with zipfile.ZipFile(result, "w", zipfile.ZIP_DEFLATED, compresslevel=9) as archive:
    for block in changed:
        archive.write(OUT / "blocks" / f"{block}.parquet", f"blocks/{block}.parquet")
        archive.write(OUT / "blocks" / f"{block}.json", f"blocks/{block}.json")
    archive.write(OUT / "manifest.json", "colab_manifest.json")
    archive.writestr("colab_environment.txt", versions + f"\ncode fingerprint {digest}\nsame-result check PASSED\n")
data = result.read_bytes()
size = bundle["result_part_mb"] * 1024 * 1024
parts = []
for i in range(0, len(data), size):
    part = Path(f"{result}.{i // size:03d}")
    part.write_bytes(data[i:i + size])
    parts.append(part)
sums = Path(f"/content/{RUN}_SHA256SUMS.txt")
sums.write_text("".join(f"{hashlib.sha256(p.read_bytes()).hexdigest()}  {p.name}\n" for p in parts)
                + f"{hashlib.sha256(data).hexdigest()}  {result.name}\n")
print([f"{p.name} {p.stat().st_size/1e6:.1f} MB" for p in parts], sums.name)
if Path("/content/drive/MyDrive").exists():
    for p in [*parts, sums]:
        shutil.copy2(p, Path("/content/drive/MyDrive") / p.name)
    print("copied to Drive")
from google.colab import files
for p in [sums, *parts]:
    files.download(str(p))
```
